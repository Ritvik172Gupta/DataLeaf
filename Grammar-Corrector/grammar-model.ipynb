```python
from transformers import T5Tokenizer, T5ForConditionalGeneration
import torch

# Load model and tokenizer
model_path = "/Users/ritvikgupta1721/Documents/MINORII/FEATURES/GRAMMAR-CORRECTOR/kaggle/working/t5_c4_grammar_corrector"  # Update if needed
tokenizer = T5Tokenizer.from_pretrained(model_path)
model = T5ForConditionalGeneration.from_pretrained(model_path)
```

```python
# Set device
device = torch.device("cuda" if torch.cuda.is_available() else "mps" if torch.backends.mps.is_available() else "cpu")
model = model.to(device)
print(device)
```

```python
def correct_grammar(text):
    input_text = "grammar: " + text
    input_ids = tokenizer.encode(input_text, return_tensors="pt", truncation=True, max_length=512).to(device)
    outputs = model.generate(input_ids, max_length=128, num_beams=4, early_stopping=True)
    corrected = tokenizer.decode(outputs[0], skip_special_tokens=True)
    return corrected
```

```python
# --- Main Program ---
def main():
    print("📝 Enter a paragraph or sentence (up to 256 words):\n")
    user_input = input("> ")

    # Word limit check
    word_count = len(user_input.strip().split())
    if word_count > 256:
        print("❌ Input exceeds 256-word limit. Please try again with shorter text.")
        return
    else:
        print(user_input)
    print("\n✅ Correcting grammar...\n")
    corrected_text = correct_grammar(user_input)
    print("🧠 Corrected Output:\n")
    print(corrected_text)
```

```python
if __name__ == "__main__":
    main()
```
